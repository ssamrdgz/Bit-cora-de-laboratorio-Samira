## 3. Gráficas

### 3.1 Primera gráfica

```python
import matplotlib.pyplot as plt
```

```python
datos = [3, 5, 7, 4, 6]
```

```python
plt.plot(datos)
plt.title("Datos simulados")
plt.xlabel("Índice")
plt.ylabel("Valor")
plt.show()
```

### 3.2 Segunda gráfica

```python
datos2 = (17, 22, 28, 23, 15, 10, 19, 20, 14, 11)
```

```python
plt.plot(datos2)
plt.title("Nueva gráfica")
plt.show()
```

### 3.3 Gráfica de barras

```python
carreras = ["Medicina", "Enfermería", "Odontología"]
estudiantes = [50, 35, 20]

plt.bar(carreras, estudiantes)
plt.title("Número de estudiantes por carrera")
plt.xlabel("Carrera")
plt.ylabel("Cantidad")

plt.show()
```

### 3.4 Gráfica de pastel

```python
frutas = ["Manzana", "Plátano", "Naranja"]
cantidad = [40, 35, 25]

plt.pie(cantidad, labels=frutas, autopct="%1.1f%%")
plt.title("Distribución de frutas")

plt.show()
```

### 3.5 Gráfica Histograma

```python
calificaciones = [60,70,80,90,75,85,95,65,70,88]

plt.hist(calificaciones)
plt.title("Distribución de calificaciones")
plt.xlabel("Calificación")
plt.ylabel("Frecuencia")

plt.show()
```

### 3.6 Gráfica de área

```python
x = [1,2,3,4,5]
y = [3,5,4,6,7]

plt.fill_between(x, y)
plt.title("Gráfica de área")
plt.xlabel("Tiempo")
plt.ylabel("Valor")

plt.show()
```

### 3.7 Gráfica de cajas o bigotes

```python
datos = [10,12,14,15,18,20,22,25,30]

plt.boxplot(datos)
plt.title("Distribución de datos")

plt.show()
```

### 3.8 Gráfica 3D

```python
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Crear datos
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

# Crear gráfica
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_title("Superficie 3D")

plt.show()
```

## 4. Mini proyecto matemático

### 4.1 Datos

```python
calificaciones = [93, 100, 98, 95, 91, 70, 96, 100]
```

### 4.2 Promedio

```python
prom = sum(calificaciones) / len(calificaciones)
print(prom)
```

### 4.3 Valor máximo

```python
c_maximo = max(calificaciones)
print(c_maximo)
```

### 4.4 Gráfica

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

plt.hist(calificaciones, bins=6)
plt.title("Distribución de calificaciones")
plt.xlabel("Calificación")
plt.ylabel("Frecuencia")
plt.grid(True)

plt.show()
```

##### La gráfica presenta intervalos de 5 en 5, lo que permite observar que la mayoría de las calificaciones se encuentran en el rango de 95 a 100 con una frecuencia de 5. En contraste, las calificaciones situadas entre 70 y 75 presentan una frecuencia de solo una vez.

# 6. Procesamiento de imagen medica

### 6.1 Imagen de retina

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("retina.jpeg")
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

### 6.2 Imagen de retina en tonos grises

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap='gray')
plt.title("Imagen en escala de grises")
plt.show()
```

### 6.3 Comparativa entre imagenes de retina

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Imagen original")

plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Escala de grises")

plt.show()
```

### 6.4 Histograma de la presencia de colores

```python
plt.hist(gray.ravel(), bins=256)
plt.title("Histograma")
plt.show()
```

### 6.5 Imagen umbralizada

```python
_, thresh = cv2.threshold(gray, 120, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(thresh, cmap='gray')
plt.title("Imagen umbralizada")
plt.show()
```

```python
cv2.imwrite("resultado_umbral.jpg", thresh)
```
