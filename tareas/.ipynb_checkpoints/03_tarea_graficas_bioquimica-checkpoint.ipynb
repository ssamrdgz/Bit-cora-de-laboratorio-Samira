##### Este proyecto utiliza diferentes tipos de gráficas para representar y comparar datos de glucosa y Hb1c obtenidos en el laboratorio. La visualización facilita la identificación de relaciones, tendencias y diferencias entre ambas variables.

### Objetivo
##### Documentar y organizar scripts de generación de gráficas para el análisis visual de datos de glucosa y Hb1c, apoyando su interpretación y comparación dentro del laboratorio.

```python
glucosa = [82, 95, 110, 130, 88, 76, 145, 99, 102, 118, 90, 104, 127, 140, 85]
hba1c = [5.1, 5.4, 5.9, 6.7, 5.2, 5.0, 7.2, 5.5, 5.8, 6.1, 5.3, 5.9, 6.6, 7.0, 5.1]
```

### 1. Histograma de glucosa

```python
import matplotlib.pyplot as plt

plt.hist(glucosa, bins=8)

plt.title("Histograma de glucosa")
plt.xlabel("Glucosa (mg/dL)")
plt.ylabel("Frecuencia")

plt.grid(True)

plt.show()
```

##### El rango más frecuente se encuentra entre 85 y 110 mg/dL, ya que es donde se encuentran los picos más altos.
##### También se observan valores altos como 127, 130, 140 y 145 mg/dL, los cuales se encuentran arriba de lo normal.

### Gráfica de barras

```python
import matplotlib.pyplot as plt

# contadores
N = 0
P = 0
D = 0

# clasificar datos
for valor in glucosa:
    if valor < 100:
        N += 1
    elif valor <= 125:
        P += 1
    else:
        D += 1

grupos = ["Normal", "Prediabetes", "Diabetes"]
conteo = [N, P, D]

# gráfica de barras
plt.bar(grupos, conteo)

plt.title("Conteo de pacientes por grupo de glucosa")
plt.xlabel("Grupo")
plt.ylabel("Número de personas")

# números arriba de cada barra
for i in range(len(conteo)):
    plt.text(i, conteo[i], conteo[i], ha='center')

plt.grid(True)

plt.show()
```

##### El grupo dominante es el de los valores normales, ya que cuenta con una mayor cantidad de datos. 

### Scatter: glucosa vs HbA1c

```python
import matplotlib.pyplot as plt

# listas por grupo
g_normal = []
a_normal = []

g_pred = []
a_pred = []

g_diab = []
a_diab = []

# clasificar puntos
for i in range(len(glucosa)):
    if glucosa[i] < 100:
        g_normal.append(glucosa[i])
        a_normal.append(hba1c[i])
    elif glucosa[i] <= 125:
        g_pred.append(glucosa[i])
        a_pred.append(hba1c[i])
    else:
        g_diab.append(glucosa[i])
        a_diab.append(hba1c[i])

# scatter por grupo
plt.scatter(g_normal, a_normal, label="Normal")
plt.scatter(g_pred, a_pred, label="Prediabetes")
plt.scatter(g_diab, a_diab, label="Diabetes")

plt.title("Relación entre glucosa y HbA1c")
plt.xlabel("Glucosa (mg/dL)")
plt.ylabel("HbA1c (%)")

plt.legend()
plt.grid(True)

plt.show()
```

##### En la gráfica se observa que mientras aumenta la glucosa, los valores de HbA1c también aumentan, lo que sugiere que existe una relación prositiva entre las variables. 

### Figura mejorada

```python
import matplotlib.pyplot as plt

# listas por grupo
g_normal = []
a_normal = []

g_pred = []
a_pred = []

g_diab = []
a_diab = []

# clasificar puntos
for i in range(len(glucosa)):
    if glucosa[i] < 100:
        g_normal.append(glucosa[i])
        a_normal.append(hba1c[i])
    elif glucosa[i] <= 125:
        g_pred.append(glucosa[i])
        a_pred.append(hba1c[i])
    else:
        g_diab.append(glucosa[i])
        a_diab.append(hba1c[i])

# tamaño de la figura
plt.figure(figsize=(7,5))

# scatter por grupo con colores y transparencia
plt.scatter(g_normal, a_normal, color="green", s=80, alpha=0.7, label="Normal")
plt.scatter(g_pred, a_pred, color="purple", s=80, alpha=0.7, label="Prediabetes")
plt.scatter(g_diab, a_diab, color="maroon", s=80, alpha=0.7, label="Diabetes")

# líneas de referencia clínicas
plt.axvline(x=100, linestyle="--", color="gray")
plt.axvline(x=126, linestyle="--", color="black")

# etiquetas completas
plt.title("Relación entre glucosa en sangre y hemoglobina glucosilada (HbA1c)")
plt.xlabel("Concentración de glucosa en sangre (mg/dL)")
plt.ylabel("Hemoglobina glucosilada HbA1c (%)")

# ajustar límites de los ejes
plt.xlim(min(glucosa)-5, max(glucosa)+5)
plt.ylim(min(hba1c)-0.5, max(hba1c)+0.5)

plt.legend()
plt.grid(True)

# guardar imagen con alta resolución
plt.savefig("grafica_glucosa_hba1c.png", dpi=300)

plt.show()
```

##### 1. La visualización de los puntos, se les cambio color, tamaño y transparencia. 
##### 2. Líneas de referencia clínicas, que permite visaulizar mejor cómo se dividen en normal, prediabetes y diabetes.
##### 3. Ajusté los límites de ejes y guardé la figura con resolución de 300 dpi, mejorando la calidad

### Líneas de referencia clínicas

```python
import matplotlib.pyplot as plt

plt.scatter(glucosa, hba1c)

# líneas de referencia clínica
plt.axvline(x=100)
plt.axvline(x=126)

plt.axhline(y=5.7)
plt.axhline(y=6.5)

plt.title("Relación entre glucosa y HbA1c con umbrales clínicos")
plt.xlabel("Glucosa (mg/dL)")
plt.ylabel("HbA1c (%)")

plt.grid(True)

plt.show()
```

##### En la gráfica con las líneas añadidas en los porcentajes y niveles de glucosa se observa que 8 muestras superan los 100 mg/dL de glucosa, indicando prediabetes o diabetes. Hasta 4 muestras llegan a superar los 120 mg/dL lo que entra en un rango de diabetes. De igual forma, 8 valores de HbA1c es mayor a 5.7% y 4 superan 6.5% lo que coincide con lo mostrado en la gráfica.
