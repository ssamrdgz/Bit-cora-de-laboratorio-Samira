### ¿Qué es?
##### Es un flujo de trabajo (pipeline) de procesamiento de imágenes comúnmente utilizado en visión artificial para mejorar la detección de bordes y características en imágenes de baja calidad, con iluminación irregular o bajo contraste.

##### *CLAHE (Contrast Limited Adaptive Histogram Equalization):* Mejora el contraste local de la imagen, haciendo que las características ocultas sean más visibles sin amplificar el ruido excesivamente.
##### *Thresholding (Thres):* Binaria la imagen, convirtiéndola en blanco y negro (0 o 255) basado en un valor umbral, lo que ayuda a segmentar objetos del fondo.
##### *Canny (Canny Edge Detector):* Algoritmo avanzado que detecta los bordes de los objetos con alta precisión, reduciendo el ruido interno mediante filtros gaussianos y siguiendo contornos

### Objetivo
##### Organizar y documentar un repositorio de procesamiento de imágenes de reticulas oculares que integre técnicas como CLAHE, detección de bordes con Canny y thresholding, con el propósito de mejorar la calidad visual de las imágenes y facilitar la delimitación de estructuras.

```python
import os
import cv2
import numpy as np
from tkinter import Tk, filedialog

# ==========================================
# 1. SELECCIONAR CARPETA DE ENTRADA Y SALIDA
# ==========================================
Tk().withdraw()  # Oculta la ventana principal de tkinter

# Elegir carpeta donde están las imágenes
carpeta_imagenes = filedialog.askdirectory(title="Selecciona la carpeta con las imágenes")

# Si no selecciona nada, salir
if not carpeta_imagenes:
    print("No se seleccionó carpeta de entrada.")
    exit()

# Elegir carpeta donde se guardarán las imágenes procesadas
carpeta_salida = filedialog.askdirectory(title="Selecciona la carpeta donde guardar las imágenes procesadas")

# Si no selecciona nada, salir
if not carpeta_salida:
    print("No se seleccionó carpeta de salida.")
    exit()

# ==========================================
# 2. CREAR OBJETO CLAHE Y PARÁMETROS
# ==========================================
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# Parámetros de Canny Edge Detection
lower_threshold = 30
upper_threshold = 70

# Parámetros para Thresholding
umbral_min = 100
umbral_max = 255

# Filtrar solo archivos de imagen
formatos_validos = (".jpg", ".png", ".jpeg")

# Lista para guardar rutas de imágenes procesadas (opcional)
imagenes_guardadas = []

# ==========================================
# 3. PROCESAR TODAS LAS IMÁGENES
# ==========================================
for archivo_imagen in os.listdir(carpeta_imagenes):
    if archivo_imagen.lower().endswith(formatos_validos):
        ruta_imagen = os.path.join(carpeta_imagenes, archivo_imagen)
        
        # Cargar la imagen en color
        imagen = cv2.imread(ruta_imagen)

        # Verificar que sí se cargó
        if imagen is None:
            print(f"No se pudo cargar la imagen: {archivo_imagen}")
            continue
        
        # Hacer una copia para no modificar la original directamente
        imagen_original = imagen.copy()
        
        # Extraer solo el canal verde
        canal_verde = imagen[:, :, 1]
        
        # Aplicar CLAHE al canal verde
        canal_verde_clahe = clahe.apply(canal_verde)
        
        # Aplicar thresholding
        _, canal_verde_umbralizado = cv2.threshold(
            canal_verde_clahe, umbral_min, umbral_max, cv2.THRESH_BINARY
        )
        
        # Aplicar Canny
        edges = cv2.Canny(canal_verde_umbralizado, lower_threshold, upper_threshold)
        
        # Convertir imágenes de 1 canal a BGR para poder unirlas
        edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
        clahe_imagen_bgr = cv2.cvtColor(canal_verde_clahe, cv2.COLOR_GRAY2BGR)
        umbralizado_imagen_bgr = cv2.cvtColor(canal_verde_umbralizado, cv2.COLOR_GRAY2BGR)
        
        # Agregar texto en cada imagen
        cv2.putText(imagen_original, 'Original: ' + archivo_imagen, (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
        
        cv2.putText(clahe_imagen_bgr, 'CLAHE (Verde)', (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
        
        cv2.putText(umbralizado_imagen_bgr, 'Umbralizado (Verde CLAHE)', (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
        
        cv2.putText(edges_bgr, 'Canny (Verde Umbralizado)', (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
        
        # Redimensionar todas las imágenes al mismo tamaño (por si acaso)
        alto = 400
        ancho = 400
        
        imagen_original = cv2.resize(imagen_original, (ancho, alto))
        clahe_imagen_bgr = cv2.resize(clahe_imagen_bgr, (ancho, alto))
        umbralizado_imagen_bgr = cv2.resize(umbralizado_imagen_bgr, (ancho, alto))
        edges_bgr = cv2.resize(edges_bgr, (ancho, alto))
        
        # Combinar horizontalmente
        imagen_combined = np.hstack((
            imagen_original,
            clahe_imagen_bgr,
            umbralizado_imagen_bgr,
            edges_bgr
        ))
        
        # Guardar la imagen combinada
        salida_path = os.path.join(carpeta_salida, f"comparacion_verde_{archivo_imagen}")
        cv2.imwrite(salida_path, imagen_combined)
        imagenes_guardadas.append(salida_path)
        
        print(f"Imagen combinada guardada en: {salida_path}")

        # ==========================================
        # 4. MOSTRAR LA IMAGEN PROCESADA
        # ==========================================
        cv2.imshow("Comparacion de Imagenes", imagen_combined)
        print("Presiona una tecla para ver la siguiente imagen...")
        cv2.waitKey(0)  # Espera hasta que presiones una tecla

# Cerrar todas las ventanas al final
cv2.destroyAllWindows()

print("Procesamiento de imágenes completado.")
```
