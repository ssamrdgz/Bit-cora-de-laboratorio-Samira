### ¿Qué es?
##### Es un flujo de trabajo (pipeline) de procesamiento de imágenes comúnmente utilizado en visión artificial para mejorar la detección de bordes y características en imágenes de baja calidad, con iluminación irregular o bajo contraste.

##### *CLAHE (Contrast Limited Adaptive Histogram Equalization):* Mejora el contraste local de la imagen, haciendo que las características ocultas sean más visibles sin amplificar el ruido excesivamente.
##### *Thresholding (Thres):* Binaria la imagen, convirtiéndola en blanco y negro (0 o 255) basado en un valor umbral, lo que ayuda a segmentar objetos del fondo.
##### *Canny (Canny Edge Detector):* Algoritmo avanzado que detecta los bordes de los objetos con alta precisión, reduciendo el ruido interno mediante filtros gaussianos y siguiendo contornos

```python
import os
import cv2
import numpy as np

# Directorio de entrada y salida
carpeta_imagenes = 'C:/Users/samir/OneDrive/Escritorio/Servicio/Bit-cora-de-laboratorio-Samira/tareas/Peorcitas'
carpeta_salida = r'C:\Users\samir\OneDrive\Escritorio\Servicio\Bit-cora-de-laboratorio-Samira\tareas\Salida_Canales_Completa'

# Crear la carpeta de salida si no existe
if not os.path.exists(carpeta_salida): #pregunta si existe
   os.makedirs(carpeta_salida) # si no existe la crea

# Crear un objeto CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# Parámetros de Canny Edge Detection
lower_threshold = 30 # umbral bajo
upper_threshold = 70 # umbral alto

# Parámetros para Thresholding, Convierte la imagen en blanco y negro puro
umbral_min = 100  # Umbral mínimo de intensidad
umbral_max = 255  # Umbral máximo de intensidad

# Filtrar solo archivos de imagen
formatos_validos = (".jpg", ".png", ".jpeg")

# Iterar sobre todas las imágenes en la carpeta de entrada
for archivo_imagen in os.listdir(carpeta_imagenes):
    if archivo_imagen.endswith(formatos_validos):
        ruta_imagen = os.path.join(carpeta_imagenes, archivo_imagen)
        
        # Cargar la imagen en color
        imagen = cv2.imread(ruta_imagen)
        
        # Extraer solo el canal verde
        canal_verde = imagen[:, :, 1]
        
        # Aplicar CLAHE al canal verde, Le mejora el contraste al canal verde
        canal_verde_clahe = clahe.apply(canal_verde)
        
        # Aplicar thresholding sobre la imagen con CLAHE, El valor del umbral usado y La imagen resultante
        _, canal_verde_umbralizado = cv2.threshold(canal_verde_clahe, umbral_min, umbral_max, cv2.THRESH_BINARY)
        
        # Aplicar Canny Edge Detection al canal verde con CLAHE, 
        # Esto intenta resaltar los contornos, por ejemplo los bordes de vasos
        edges = cv2.Canny(canal_verde_umbralizado, lower_threshold, upper_threshold)
        
        # Convertir la imagen de bordes a BGR para combinarla con las otras
        edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR) # edges está en una sola capa (grises)
        # Pero la imagen original tiene 3 canales (BGR).Para poder unirlas bien con np.hstack, todas deben tener la misma forma
        # Entonces conviertes la imagen de grises a “falso color” BGR
        
        # Convertir el canal verde CLAHE a BGR para visualizarlo
        clahe_imagen_bgr = cv2.cvtColor(canal_verde_clahe, cv2.COLOR_GRAY2BGR)
        
        # Convertir el canal verde umbralizado a BGR para visualizarlo
        umbralizado_imagen_bgr = cv2.cvtColor(canal_verde_umbralizado, cv2.COLOR_GRAY2BGR)
        
        # Agregar el nombre en la imagen original
        cv2.putText(imagen, 'Original: ' + archivo_imagen, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        # Agregar el nombre en la imagen CLAHE del canal verde
        cv2.putText(clahe_imagen_bgr, 'CLAHE (Verde): ' + archivo_imagen, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        # Agregar el nombre en la imagen umbralizada
        cv2.putText(umbralizado_imagen_bgr, 'Umbralizado (Verde CLAHE): ' + archivo_imagen, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        # Agregar el nombre en la imagen de bordes
        cv2.putText(edges_bgr, 'Canny (Verde Umbralizado): ' + archivo_imagen, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        # Combinar la imagen original, CLAHE del canal verde, umbralizado y bordes horizontalmente
        imagen_combined = np.hstack((imagen, clahe_imagen_bgr, umbralizado_imagen_bgr, edges_bgr))
        
        # Guardar la imagen combinada en la carpeta de salida
        salida_path = os.path.join(carpeta_salida, f"comparacion_verde_{archivo_imagen}")
        cv2.imwrite(salida_path, imagen_combined)
        print(f"Imagen combinada guardada en: {salida_path}")

print("Procesamiento de imágenes completado.")
```

```python
# Crear la carpeta de salida si no existe
if not os.path.exists(carpeta_salida):
    os.makedirs(carpeta_salida)

# Crear un objeto CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# Parámetros de Canny Edge Detection
lower_threshold = 30
upper_threshold = 70

# Parámetros para Thresholding
umbral_min = 100  # Umbral mínimo de intensidad
umbral_max = 255  # Umbral máximo de intensidad

# Filtrar solo archivos de imagen
formatos_validos = (".jpg", ".png", ".jpeg")

# Iterar sobre todas las imágenes en la carpeta de entrada
for archivo_imagen in os.listdir(carpeta_imagenes):
    if archivo_imagen.lower().endswith(formatos_validos):
        ruta_imagen = os.path.join(carpeta_imagenes, archivo_imagen)
        
        # Cargar la imagen en color
        imagen = cv2.imread(ruta_imagen)
        
        # Separar canales BGR
        b, g, r = cv2.split(imagen)
        canales = {'Azul': b, 'Verde': g, 'Rojo': r}
        
        resultados = []

        # Procesar cada canal
        for nombre, canal in canales.items():
            
            # CLAHE
            canal_clahe = clahe.apply(canal)
            
            # Threshold
            _, canal_umbral = cv2.threshold(canal_clahe, umbral_min, umbral_max, cv2.THRESH_BINARY)
            
            # Canny
            edges = cv2.Canny(canal_umbral, lower_threshold, upper_threshold)
            
            # Convertir a BGR para visualizar
            clahe_bgr = cv2.cvtColor(canal_clahe, cv2.COLOR_GRAY2BGR)
            umbral_bgr = cv2.cvtColor(canal_umbral, cv2.COLOR_GRAY2BGR)
            edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
            
            # Agregar títulos
            cv2.putText(clahe_bgr, f'CLAHE ({nombre})', (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
            
            cv2.putText(umbral_bgr, f'Umbral ({nombre})', (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
            
            cv2.putText(edges_bgr, f'Canny ({nombre})', (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
            
            # Guardar resultados
            resultados.extend([clahe_bgr, umbral_bgr, edges_bgr])

        # Agregar título a la imagen original
        cv2.putText(imagen, 'Original', (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

        # Combinar todo
        imagen_combined = np.hstack([imagen] + resultados)
        
        # Guardar
        salida_path = os.path.join(carpeta_salida, f"comparacion_{archivo_imagen}")
        cv2.imwrite(salida_path, imagen_combined)
        print(f"Imagen combinada guardada en: {salida_path}")

print("Procesamiento de imágenes completado.")
```
