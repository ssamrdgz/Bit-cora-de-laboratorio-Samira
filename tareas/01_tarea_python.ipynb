## Operaciones algebraicas

```python
x = 8
y = 3
```

```python
suma = [x + y]
print(suma)
```

```python
resta = [x - y]
print(resta)
```

```python
multiplicacion = [x * y]
print(multiplicacion)
```

```python
division = [x / y]
print(division)
```

```python
cuadrados = [x**2 + y**2]
print(cuadrados)
```

## Promedio y operaciones con listas

```python
lista = [85, 90, 78, 92, 88]
```

```python
promedio = sum(lista) / len(lista)
print(promedio)
```

```python
maximo = max(lista)
print(maximo)
```

```python
minimo = min(lista)
print(minimo)
```

```python
rango = max(lista) - min(lista)
print(rango)
```

## Introduccion a vectores

```python
v1 = [1, 2, 3]
v2 = [4, 5, 6]
```

```python
sumav = [v1[i] + v2[i] for i in range(len(v1))]
print(sumav)
```

```python
restav = [v1[i] - v2[i] for i in range(len(v1))]
print(restav)
```

```python
producto = [v1[i] * v2[i] for i in range(len(v1))]
print(producto)
```

## ¿Qué relación encuentras entre las operaciones algebraicas y las operaciones que realizamos en Python?

##### Las operaciones algebraicas y las operaciones en Python siguen las mismas reglas matemáticas. La diferencia es que Python es un sistema computacional que permite realizar las operaciones de una manera más sintetizada.
