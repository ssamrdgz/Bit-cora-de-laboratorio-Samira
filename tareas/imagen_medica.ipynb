# Procesamiento de imagen

### Imagen de placa

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("placa.jpeg")
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

### Imagen en tonos grises 

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap='gray')
plt.title("Imagen en escala de grises")
plt.show()
```

### Comparativo de imagenes

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Imagen original")

plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Escala de grises")

plt.show()
```

### Hisograma de los tonos de grises en la imagen

```python
plt.hist(gray.ravel(), bins=256)
plt.title("Histograma")
plt.show()
```

##### En el histograma se aprecía una mayor frecuencia entre los valores 60 a 125, lo que indica que en la imagen predominan los tonos grises medios. Mientras que se observa en 0 no hay presencia de nada, debido a que no hay presencia del negro en la imagen y los tonos por 250 son muy minímos, debido a que tampoco hay presencia de blancos.

### Imagen umbralizada de la placa

```python
_, thresh = cv2.threshold(gray, 130, 255, cv2.THRESH_BINARY)
```

```python
plt.imshow(thresh, cmap='gray')
plt.title("Imagen umbralizada")
plt.show()
```

```python
cv2.imwrite("resultado_umbral.jpg", thresh)
```

### Preguntas de análisis

#### ¿Qué diferencias observas entre la imagen original y la imagen en escala de grises?

##### La imagen en escala de grises permite observar mayores detalles que se pierden en la imagen original por los colores. 

#### ¿Qué representa el histograma de una imagen?

##### El histograma representa la frecuencia o la distribución de los pixeles en la imagen en una escala de tonos grises, mostrando cuantos pixeles hay entre 0 a 255, siendo el 0 negro y 255 el blanco. 

#### ¿Qué efecto tiene la umbralización sobre la imagen?

##### La imagen umbralizada permite aumentar el contraste entre las zonas blanca y negras, convirtiendo la imagen en tonos grises en una imagen binaria y resaltando las zonas de interés.

#### ¿Cómo podría utilizarse esta técnica en el análisis de imágenes médicas?

##### El uso de la imagen umbralizada puede permitir separar estructuras del fondo o de otros tejidos, detectar anomalías o hasta preparar la imagen para estudios automatizados.
