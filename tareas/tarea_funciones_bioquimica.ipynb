### Funciones estadísticas básicas

```python
def estadisticas_basicas(lista):
    media = sum(lista)/len(lista)
    minimo = min(lista)
    maximo = max(lista)
    lista_ordenada = sorted(lista)
    n = len(lista_ordenada)
    if n % 2 == 0:
        mediana = (lista_ordenada[n//2 - 1] + lista_ordenada[n//2]) / 2
    else:
        mediana = lista_ordenada[n//2]
        return media, minimo, maximo, mediana
```

#### Resultados glucosa

```python
resultado_g = estadisticas_basicas(glucosa)
print(resultado_g)
```

#### Resultados hba1c

```python
resultado_h = estadisticas_basicas(hba1c)
print(resultado_h)
```

### Función de clasificación clínica

```python
def clasificar_glucosa(valor):
    
    if valor < 100:
        return "Normal"
    elif valor < 126:
        return "Prediabetes"
    else:
        return "Diabetes"
```

```python
clasificar_glucosa(146)
```

```python
clasificar_glucosa(94)
```

```python
clasificar_glucosa(117)
```

```python
def clasificar_lista(lista):
    
    normales = 0
    prediabetes = 0
    diabetes = 0
    for valor in lista:
        categoria = clasificar_glucosa(valor) 
        if categoria == "Normal":
            normales += 1
        elif categoria == "Prediabetes":
            prediabetes += 1
        else:
            diabetes += 1
    return normales, prediabetes, diabetes
```

```python
resultado = clasificar_lista(glucosa)

print("Normales:", resultado[0])
print("Prediabetes:", resultado[1])
print("Diabetes:", resultado[2])
```

### Correlación simple

```python
import math
def correlacion_simple(x, y):
    n = len(x)
    # Medias
    media_x = sum(x) / n
    media_y = sum(y) / n
    # Numerador
    numerador = 0
    for i in range(n):
        numerador += (x[i] - media_x) * (y[i] - media_y)
    # Denominador
    suma_x = 0
    suma_y = 0
    for i in range(n):
        suma_x += (x[i] - media_x) ** 2
        suma_y += (y[i] - media_y) ** 2
    denominador = math.sqrt(suma_x * suma_y)
    r = numerador / denominador
    return r
```

```python
r = correlacion_simple(glucosa, hba1c)
print(r)
```

### Intento de gráfica

```python
def grafica_histograma(datos, titulo, xlabel):
    
    print("\n" + titulo)
    print(xlabel)
    print("-" * 30)
    
    minimo = min(datos)
    maximo = max(datos)
    
    bins = 5
    ancho = (maximo - minimo) / bins
    
    frecuencias = [0] * bins
    
    # Contar frecuencias
    for valor in datos:
        indice = int((valor - minimo) / ancho)
        
        if indice == bins:  # evitar error en el valor máximo
            indice -= 1
            
        frecuencias[indice] += 1
    
    # Mostrar histograma
    for i in range(bins):
        limite_inf = minimo + i * ancho
        limite_sup = limite_inf + ancho
        
        print(f"{limite_inf:.2f} - {limite_sup:.2f} | {'*' * frecuencias[i]}")
```

```python
grafica_histograma(glucosa, "Histograma Glucosa", "Glucosa (mg/dL)")
```
